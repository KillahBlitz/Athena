
# Conversion de PDF a Markdown

Este script extrae la informacion de los programas academicos de PDFs a Markdown para la extraccion de texto.
El objetivo de este script es disminuir el consumo de tokens para la extraccion de texto y normalizacion de campos.


### Requisitos
- instalar markitdown y dependencias de biblioteca con entorno virtual
- descargar programas academicos

```python
import os

class ExtractorPDF:
    def __init__(self, route_pdfs: str, route_md: str):
        self.route_pdfs = route_pdfs
        self.route_md = route_md

    def run_command_bash(self):
        pdf_files = [f for f in os.listdir(self.route_pdfs) if f.endswith('.pdf')]
        for pdf_file in pdf_files:
            md_file = os.path.splitext(pdf_file)[0] + '.md'
            command = f"markitdown '{os.path.join(self.route_pdfs, pdf_file)}' > '{os.path.join(self.route_md, md_file)}'"
            os.system(command)
            print(f"Extracted text from {pdf_file} to {md_file}")
```

```python
if __name__ == "__main__":
    route_pdfs = "/Users/jmonroy/Documents/MyProjects/Athena/Scripts/python/assets/layouts/pdf"
    route_md = "/Users/jmonroy/Documents/MyProjects/Athena/Scripts/python/assets/layouts/test"

    extractor = ExtractorPDF(route_pdfs, route_md)
    extractor.run_command_bash()
```
